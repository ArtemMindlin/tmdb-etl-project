```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
PROCESSED_DATA_PATH = Path("../data/processed/")

df_genres = pd.read_csv(PROCESSED_DATA_PATH / "tmdb_genres.csv")
df_popular = pd.read_csv(PROCESSED_DATA_PATH / "tmdb_popular.csv", engine='python')
df_top_rated = pd.read_csv(PROCESSED_DATA_PATH / "tmdb_top_rated.csv")
df_upcoming = pd.read_csv(PROCESSED_DATA_PATH / "tmdb_upcoming.csv")
```

```python
df_popular.isna().sum().plot(kind="bar", figsize=(12, 4))
plt.title("Missing Values per Column")
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df_top_rated["vote_average"], bins=30)

plt.title("Distribution of Movie Popularity")
plt.xlabel("Popularity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
df_popular.groupby("year")["popularity"].mean().plot()
```
